```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

path = r"C:\Users\ARYAN MAHAKUR\Desktop\VS CODE\SAFEDEV.AI\ProcessedDataSet.csv"

df = pd.read_csv(path)

print("Shapes:", df.shape)
print("First 5 rows:", df.head())
print("Columns:", df.columns.tolist())
print("Data types:", df.dtypes)
print("Missing values:", df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Unique values:", df.nunique())
print("Statistics:", df.describe())
print("Infinite values:", np.isinf(df.select_dtypes(include=np.number)).sum())
print("Attack Type:", df["Label"].value_counts())
print("Attack Type Percentage:", df["Label"].value_counts(normalize=True) * 100)

summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing": df.isnull().sum(),
    "Unique": df.nunique()
})

print("Summary:", summary)
```
